```python
import pandas as pd
import numpy as np
import plotly.express as px

# --- 1. Gráfico de barras ---
# Datos categóricos de ejemplo
df_cat = pd.DataFrame({
    'Categoria': ['A', 'B', 'C', 'D', 'E'],
    'Valores': [23, 45, 56, 78, 12]
})
# Crear gráfico de barras
fig_bar = px.bar(df_cat, x='Categoria',  y='Valores', title='Gráfico de Barras - Valores por Categoría', labels={'Valores':'Cantidad', 'Categoria':'Grupo'})
fig_bar.show()

# --- 2. Gráfico de líneas (serie temporal) ---
# Generar fechas mensuales y valores aleatorios
fechas = pd.date_range(start='2024-01-01', periods=12, freq='M')
df_time = pd.DataFrame({
    'Fecha': fechas,
    'Ventas': np.random.randint(100, 500, size=len(fechas))
})
# Crear gráfico de líneas
fig_line = px.line(
    df_time,
    x='Fecha',
    y='Ventas',
    title='Gráfico de Líneas - Ventas Mensuales',
    markers=True
)
fig_line.update_layout(xaxis_title='Mes', yaxis_title='Ventas (€)')
fig_line.show()

# --- 3. Gráfico de dispersión (scatter) ---
# Usar dataset 'iris' incorporado
df_iris = px.data.iris()
fig_scatter = px.scatter(
    df_iris,
    x='sepal_width',
    y='sepal_length',
    color='species',
    title='Scatter Plot - Iris (sépalos)',
    labels={'sepal_width':'Anchura Sépalo', 'sepal_length':'Longitud Sépalo'}
)
fig_scatter.show()

# --- 4. Gráfico 3D de dispersión ---
fig_3d = px.scatter_3d(
    df_iris,
    x='sepal_width',
    y='sepal_length',
    z='petal_length',
    color='species',
    title='Scatter 3D - Iris (sépalos/pétalos)'
)
fig_3d.show()

# --- 5. Gráfico de pastel (pie) ---
fig_pie = px.pie(
    df_cat,
    values='Valores',
    names='Categoria',
    title='Gráfico de Pastel - Distribución de Categorías'
)
fig_pie.show()

# --- 6. Histograma ---
# Datos aleatorios normalmente distribuidos
df_hist = pd.DataFrame({'Datos': np.random.randn(500)})
fig_hist = px.histogram(
    df_hist,
    x='Datos',
    nbins=30,
    title='Histograma - Distribución Normal',
    labels={'Datos':'Valor'}
)
fig_hist.show()

# --- 7. Box Plot ---
# Usar dataset 'tips' incorporado para gráfico de caja
df_tips = px.data.tips()
fig_box = px.box(
    df_tips,
    x='day',
    y='total_bill',
    color='smoker',
    title='Box Plot - Cuenta Total por Día y Fumador',
    labels={'day':'Día', 'total_bill':'Cuenta (€)', 'smoker':'¿Fumador?'}
)
fig_box.show()
```
